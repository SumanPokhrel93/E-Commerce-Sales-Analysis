## Key Findings

- The business generated a total revenue of 5.28 million and a total profit of 1.44 million from 5,348 unique customers.
- The Middle East was the highest revenue-generating region, followed by North America and Asia.
- Electronics was the top-performing category, contributing the highest revenue and profit among all product categories.
- Regular customers generated the highest overall profit, highlighting the importance of customer retention.
- Technology products such as Lenovo ThinkPad X1 and MacBook Air M2 were among the highest revenue-generating products.
- Revenue showed strong growth from 2021 to 2023 before declining significantly during 2024.

```python
import matplotlib.pyplot as plt
```

```python
Rev_Region = df.groupby('Region')['Revenue'].sum().sort_values(ascending=False)
Rev_Region.plot(kind='bar', figsize=(10,5))
plt.title ('Revenue by Region')
plt.xlabel('Region')
plt.ylabel('Revenue generated (in millions)')
plt.savefig('Revenue by Region', bbox_inches='tight')
plt.show()
```

```python
Rev_Category = df.groupby('Category')['Revenue'].sum().sort_values(ascending=False)
Rev_Category.plot(kind='bar', figsize = (10,5))
plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Revenue generated (in millions)')
plt.savefig('Revenue by Category', bbox_inches='tight')
plt.show()
```

```python
Ship_Cost = df.groupby('Shipping_Method')['Shipping_Cost'].sum().sort_values(ascending=False)
Ship_Cost.plot(kind='bar', figsize =(10,5))
plt.title('Shipping costs of different Methods')
plt.xlabel('Shipping Method')
plt.ylabel('Total Cost of Shipping')
plt.savefig('Shipping Cost of different Methods', bbox_inches='tight')
plt.show()
```

```python
Customer_Seg = df.groupby('Customer_Segment')['Profit'].sum().sort_values(ascending=False)
Customer_Seg.plot(kind='bar', figsize=(10,5))
plt.title('Profit Generated by different Customer Segments')
plt.xlabel('Type of Customer')
plt.ylabel('Profit generated')
plt.savefig('Profit By Customer Segments', bbox_inches='tight')
plt.show()
```

```python
df['YearMonth'] = df['Order_Date'].dt.to_period('M')
Monthly_rev = df.groupby('YearMonth')['Revenue'].sum()
Monthly_rev.plot(kind='line', figsize=(10,5))
plt.title('Monthly Revenue Trend')
plt.ylabel('Revenue')
plt.savefig('Monthly Revenue Trend line', bbox_inches='tight')
plt.show()
```

```python
Best_Products = df.groupby('Product_Name')['Revenue'].sum().sort_values(ascending=False).head(10)
Best_Products.plot(kind='bar', figsize = (10,5))
plt.title('Highest Revenue generating Products')
plt.xlabel('Name of Product')
plt.ylabel('Total Revenue Generated')
plt.savefig('Best Performing Products')
plt.show()
```

## Conclusion

- The business generated over 5.28 million in revenue and 1.44 million in profit from 5,348 unique customers.
- The Middle East was the highest revenue-generating region, indicating strong market performance.
- Electronics was the best-performing category, contributing the highest revenue and profit.
- Regular customers generated the highest overall profit, highlighting the importance of customer retention.
- Revenue grew steadily from 2021 to 2023 before showing a decline in 2024, suggesting an area for further business investigation.
